```python
import numpy as np
import matplotlib.pyplot as plt
from collections import Counter
```

```python
data = np.random.rand(100)
```

```python
train = data[:50]
labels = ["class1" if x <=0.5 else "class2" for x in train]

test = data[50:]

def knn(train,labels,point, k):
  d = sorted([(abs(point - train[i]), labels[i]) for i in range(len(train))])
  return Counter([label for _, label in d[:k]]).most_common(1)[0][0]

k_values =[1,2,3,45,20,30]

for k in k_values:
  pred = [knn(train,labels,x, k) for x in test]

  print(f"k= {k} :")
  for i , p in enumerate(pred,51):
    print(f"x{i} ({test[i-51]:.4f})-> {p}")

plt.figure()
plt.scatter(train,[0]*50, c=["blue" if l=="class1" else "red" for l in labels])
plt.scatter(test, [1]*50, c=["blue" if p=="class1" else "red" for p in pred])
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.preprocessing import PolynomialFeatures, StandardScaler
from sklearn.datasets import fetch_california_housing
from sklearn.pipeline import make_pipeline
```

```python
df = fetch_california_housing(as_frame=True).frame
X = df[['AveRooms']]
y = df['MedHouseVal']

Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=0.2)

lr = LinearRegression().fit(Xtr, ytr)
yp = lr.predict(Xte)

plt.scatter(Xte, yte)
plt.plot(Xte, yp, color='red')
plt.title("linear Regression")
plt.show()

print(mean_squared_error(yte, yp))
print(r2_score(yte, yp))


url = "https://archive.ics.uci.edu/ml/machine-learning-databases/auto-mpg/auto-mpg.data"
cols=["mpg","cyl","disp","hp","wt","acc","yr","org"]

data = pd.read_csv(url, sep='\s+', names=cols, na_values="?").dropna()

X = data[['disp']]
y = data['mpg']

Xtr, Xte,ytr, yte = train_test_split(X, y ,  test_size=0.2)

model= make_pipeline(PolynomialFeatures(2), StandardScaler(), LinearRegression())
model.fit(Xtr,ytr)
yp=model.predict(Xte)

plt.scatter(Xte, yte)
plt.scatter(Xte,yp,color='red')
plt.show()

print(mean_squared_error(yte,yp))
print(r2_score(yte,yp))


```
